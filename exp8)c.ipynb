```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
# The sentence you want to encode
sentence = "I am staying in Vijayawada."
# 1. Tokenize the sentence into a list of words
words = sentence.lower().replace('.', '').split()
print(f"Tokenized words: {words}")
print("-" * 30)
# 2. Create a DataFrame from the list of words
# We reshape it to a column vector because fit_transform expects a 2D array
df = pd.DataFrame(words, columns=['word'])
print("DataFrame from words:")
print(df)
print("-" * 30)
# 3. Initialize the OneHotEncoder
encoder = OneHotEncoder(sparse_output=False)
# 4. Fit and transform the data
encoded_words = encoder.fit_transform(df[['word']])
# 5. Get the feature names (the words themselves)
feature_names = encoder.get_feature_names_out(['word'])
# 6. Create a new DataFrame with the one-hot encoded result
one_hot_df = pd.DataFrame(encoded_words, columns=feature_names)
print("One-Hot Encoded DataFrame:")
print(one_hot_df)
```
